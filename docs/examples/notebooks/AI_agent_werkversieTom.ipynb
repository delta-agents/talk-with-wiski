Voorbeeld hoe je de example tool gebruikt. 

## Parameters

Met de tool `get_parameters` kan je de parameters van WISKI opvragen.

```python
from talk_with_wiski.models import GetLocationsInput, GetParametersInput, GetTimeseriesListInput, GetTimeseriesValuesInput
from talk_with_wiski.tools import get_locations, get_parameters, get_timeseries_list, get_timeseries_values
from setup_quarto import setup_dot_env
import json
```

```python
ctx = setup_dot_env()
```

```python
import numpy as np

example_input = GetParametersInput()
example_output = await get_parameters(ctx, example_input)

datadict = example_output.data
# datastr = json.dumps(datadict, indent=2)
parameters = np.unique(
    [row["stationparameter_longname"].split(" [")[0].lower() for row in datadict]
)
print(parameters)
```

### Using geocoding is also possible

This uses `from geopy.geocoders import Nominatim`, <br>
which uses the OSM API, [See the policy](https://operations.osmfoundation.org/policies/nominatim/)<br>
 -> `No heavy uses (an absolute maximum of 1 request per second).` 

```python
input = GetLocationsInput(
    requested_geolocation=["Breda", 5.0],
    returnfields="station_no,station_id,station_name,stationparameter_name,station_latitude,station_longitude,site_name",
)
print(input)
output = await get_locations(ctx, input)
for item in output.data:
    print(json.dumps(item, indent=2))
print(f"Aantal records returned {len(output.data)}")
```

```python
# example_input = GetParametersInput(
#     station_no=["ZUN0034_KKL_OW", "0300_TDB"]
# )
example_input = GetParametersInput(
    station_no=output.data[4]["station_no"],
    # parameter_type='debiet'
)
example_output = await get_parameters(ctx, example_input)

datadict_parmaters = example_output.data
datastr = json.dumps(datadict_parmaters, indent=2)
print(datastr)
```

## Timeseries list

Met de tool `get_timeseries_list` kan je de beschikbare timeseries ophalen voor een locatie, site en/of een parameter.

```python
example_input = GetTimeseriesListInput(
    station_no=datadict_parmaters[0]["station_no"],
    parametertype_name=[item["parametertype_name"] for item in datadict_parmaters]
)
example_output = await get_timeseries_list(ctx, example_input)

datadict_ts = example_output.data
datastr = json.dumps(datadict_ts, indent=2)
print(datastr)
```

## Timeseries values

Met de tool `get_timeseries_values` kan je de timeseries data ophalen voor een timeseries_id.

```python
example_input = GetTimeseriesValuesInput(
    ts_id=[item["ts_id"] for item in datadict_ts],
    from_date="2025-11-01",
    period="P1D",
    # to_date="2025-11-06"
)
example_output = await get_timeseries_values(ctx, example_input)

datadict_timeseries = example_output.data
datastr = json.dumps(datadict_timeseries, indent=2)
print(datastr)
```
